### 1.2 Conversion et chargement du fichier en CSV

Le fichier fourni est en Excel. On le convertit une seule fois en CSV car il se lit beaucoup plus vite. Les identifiants sont lus comme du texte et la date comme une date. Les noms de colonnes sont traduits en français. `df_raw` garde les données brutes et le nettoyage se fait sur la copie `df`.

```python
import os

# Conversion unique Excel -> CSV
if not os.path.exists("../data/online_retail.csv"):
    pd.read_excel("../data/Online_Retail.xlsx", dtype={"InvoiceNo": str, "StockCode": str}).to_csv("../data/online_retail.csv", index=False)

# Chargement
df_raw = pd.read_csv("../data/online_retail.csv", dtype={"InvoiceNo": str, "StockCode": str}, parse_dates=["InvoiceDate"])

# Traduction des noms de colonnes
df_raw = df_raw.rename(columns={
    "InvoiceNo": "numero_facture",
    "StockCode": "code_produit",
    "Description": "description",
    "Quantity": "quantite",
    "InvoiceDate": "date_facture",
    "UnitPrice": "prix_unitaire",
    "CustomerID": "id_client",
    "Country": "pays"
})

# Copie de travail
df = df_raw.copy()
```

### 2.7 Codes produits particuliers

On a croisé des codes produits avec lettre ("D" pour Discount, "B" pour bad debt). Les codes normaux commencent par 5 chiffres (ex. 85123A). On liste ceux qui ne suivent pas ce format.

```python
# On cherche les codes produits qui ne commencent pas par 5 chiffres, avec leur description et leur nombre de lignes
codes_speciaux = df[~df["code_produit"].str.match(r"^\d{5}")]

codes_speciaux.groupby("code_produit").agg(
    description=("description", "first"),
    nb_lignes=("code_produit", "size")
).sort_values("nb_lignes", ascending=False)
```

Les codes qui ne commencent pas par 5 chiffres se rangent en plusieurs familles :
- **Frais de livraison** : POST, DOT, C2 (carriage). Ce sont des frais d'envoi, pas des produits.
- **Frais et commissions** : BANK CHARGES, AMAZONFEE, CRUK (commission versée à une association). Ce sont des coûts, pas des achats.
- **Ajustements** : D (remise), B (bad debt), S (échantillons), M et m (saisie manuelle, on ne sait pas ce qu'il y a derrière).
- **Bons cadeaux** : gift_0001_XX. C'est un vrai achat mais d'un bon et pas d'un produit.
- **Vrais produits avec un code non standard** : les DCGS (sacs de fête, colliers pour chien...) et PADS. Quelques DCGS n'ont pas de description ou juste "ebay", donc plutôt des ajustements.

Le format "5 chiffres" ne suffit pas pour repérer les intrus car il attrape aussi de vrais produits. On triera au nettoyage.

```python
# Les frais de livraison sont-ils facturés au client ? On regarde leur prix et s'ils ont un client
livraison = df[df["code_produit"].isin(["POST", "DOT", "C2"])]

livraison.groupby("code_produit").agg(
    nb_lignes=("prix_unitaire", "size"),
    prix_min=("prix_unitaire", "min"),
    prix_median=("prix_unitaire", "median"),
    prix_max=("prix_unitaire", "max"),
    sans_client=("id_client", lambda x: x.isna().sum())
)
```

J'ai vérifié les frais de livraison (POST, DOT, C2) : ils ont un prix positif sur la facture du client donc c'est bien le client qui les paie (le port est parfois offert, prix à 0, mais c'est rare). Les DOT n'ont presque jamais de client et partiront au nettoyage. Pour POST et C2, il faudra décider si on les garde dans le montant : c'est de l'argent dépensé, mais pas un achat de produit. Quelques prix max sont aberrants (8 142 pour un POST).

### 2.11 Annulations et achats d'origine

Les valeurs extrêmes ont montré que les annulations ne suppriment pas l'achat d'origine : la commande reste dans les données. On vérifie si c'est le cas pour toutes les annulations.

```python
# Pour chaque annulation, on cherche l'achat d'origine : même client, même produit, même quantité en positif
annul = df[annulations & df["id_client"].notna()].copy()
achats = df[~annulations & df["id_client"].notna()]

annul["quantite_positive"] = -annul["quantite"]

correspondances = annul.merge(
    achats[["id_client", "code_produit", "quantite"]].drop_duplicates(),
    left_on=["id_client", "code_produit", "quantite_positive"],
    right_on=["id_client", "code_produit", "quantite"],
    how="left",
    indicator=True
)

print("Annulations avec client :", len(annul))
correspondances["_merge"].value_counts()
```

```python
# On regarde quelques annulations sans achat correspondant pour comprendre pourquoi on ne les retrouve pas
correspondances[correspondances["_merge"] == "left_only"].head(10)
```

```python
# On classe les annulations sans achat correspondant selon la raison probable
sans_achat = correspondances[correspondances["_merge"] == "left_only"]

# Les paires client + produit qu'on trouve dans les achats (peu importe la quantité)
paires_achetees = achats[["id_client", "code_produit"]].drop_duplicates()
sans_achat = sans_achat.merge(paires_achetees, on=["id_client", "code_produit"], how="left", indicator="deja_achete")

code_special = ~sans_achat["code_produit"].str.match(r"^\d{5}")
produit_achete = sans_achat["deja_achete"] == "both"

print("Code particulier (remise, saisie manuelle...) :", code_special.sum())
print("Produit acheté mais quantité différente (annulation partielle) :", (~code_special & produit_achete).sum())
print("Produit jamais acheté dans le dataset (achat hors période) :", (~code_special & ~produit_achete).sum())
```

```python
# Répartition des annulations avec client selon leur type
types_annulations = pd.Series({
    "Totales": (correspondances["_merge"] == "both").sum(),
    "Partielles": (~code_special & produit_achete).sum(),
    "Hors période": (~code_special & ~produit_achete).sum(),
    "Codes particuliers": code_special.sum()
})

types_annulations.plot(kind="bar", figsize=(8, 4), rot=0, color=["tab:green", "tab:orange", "tab:red", "tab:gray"])
plt.title("Types d'annulations (avec client)")
plt.ylabel("Nombre d'annulations")
plt.show()
```

Il y a 8 905 annulations avec un client.
- 3 266 retrouvent leur achat d'origine à l'identique (même client, même produit et même quantité). Ce sont des annulations totales.
- 4 670 portent sur un produit que le client a bien acheté mais pas dans la même quantité. Ce sont surtout des retours partiels.
- 693 portent sur un produit jamais acheté dans nos données. L'achat d'origine date sûrement d'avant décembre 2010.
- 276 sont des codes particuliers (remises et saisies manuelles).

Le graphique montre que les retours partiels sont les plus nombreux (52 % des annulations), devant les annulations totales (37 %). Ensemble ils représentent presque 9 annulations sur 10 dont l'achat d'origine est encore dans les données.

Supprimer seulement les lignes "C" laisserait donc ces achats annulés ou retournés dans les données. Ça gonflerait le montant de beaucoup de clients. Il faudra en tenir compte au nettoyage.

## 3. Nettoyage des données

On applique les décisions prises dans la synthèse de l'EDA, dans cet ordre. À chaque étape, on note le nombre de lignes supprimées.

### 3.1 Suppression des lignes sans id client

```python
# On supprime les lignes sans id client, puis on passe l'id en entier (plus de NaN, donc plus besoin du float)
avant = len(df)

df = df.dropna(subset=["id_client"])
df["id_client"] = df["id_client"].astype(int)

print(f"Lignes supprimées : {avant - len(df)}")
print(f"Lignes restantes : {len(df)}")
```

135 080 lignes supprimées, il en reste 406 829.

### 3.2 Suppression des doublons

```python
# On supprime les lignes strictement identiques (on garde la première occurrence)
avant = len(df)

df = df.drop_duplicates()

print(f"Lignes supprimées : {avant - len(df)}")
print(f"Lignes restantes : {len(df)}")
```

5 225 doublons supprimés (les 43 autres n'avaient pas d'id client et sont partis à l'étape d'avant). Il reste 401 604 lignes.

### 3.3 Suppression des codes non-produits

On retire les lignes qui ne sont pas des achats de produits : frais de port, frais et commissions, ajustements et bons cadeaux. Les DCGS et PADS sont de vrais produits, on les garde.

```python
# On supprime les codes qui ne sont pas des produits (liste précise, pour ne pas toucher aux DCGS et PADS)
codes_a_retirer = ["POST", "DOT", "C2", "BANK CHARGES", "AMAZONFEE", "CRUK", "M", "m", "B", "D", "S"]
est_non_produit = df["code_produit"].isin(codes_a_retirer) | df["code_produit"].str.startswith("gift_0001")

avant = len(df)
print(df.loc[est_non_produit, "code_produit"].value_counts())

df = df[~est_non_produit]

print(f"\nLignes supprimées : {avant - len(df)}")
print(f"Lignes restantes : {len(df)}")
```

1 911 lignes supprimées, surtout des frais de port (POST, C2) et des saisies manuelles (M). Les autres codes de la liste (échantillons, frais Amazon, bons cadeaux...) n'avaient pas d'id client et étaient déjà partis. Il reste 399 693 lignes.

### 3.9 Dataset nettoyé pour l'analyse (df_clean)

`df` contient encore les retours partiels en lignes négatives. Pour les analyses de Data Studio on crée `df_clean` où chaque retour est fusionné avec l'achat du même client et du même article. Par exemple 10 tasses achetées puis 5 retournées deviennent une seule ligne de 5 tasses. On retire aussi les clients dont le total net est nul ou négatif. Ce sont les mêmes que ceux retirés en 4.3.

`df_clean` ne sert plus jusqu'à l'export de la partie 12.

```python
# On fusionne chaque retour avec les achats du même client et du même article, puis on retire les clients au total net nul ou négatif
df_clean = df.sort_values("date_facture").copy()
df_clean["cle"] = df_clean["id_client"].astype(str) + "_" + df_clean["code_produit"]

# Seules les paires client + article qui ont au moins un retour sont concernées
cles_retour = df_clean.loc[df_clean["quantite"] < 0, "cle"].unique()
a_traiter = df_clean[df_clean["cle"].isin(cles_retour)]

quantites = df_clean["quantite"].to_dict()

for _, groupe in a_traiter.groupby("cle"):
    achats_passes = []
    for idx, qte in groupe["quantite"].items():
        if qte > 0:
            achats_passes.append(idx)
        else:
            reste = -qte
            for idx_achat in reversed(achats_passes):
                pris = min(quantites[idx_achat], reste)
                quantites[idx_achat] -= pris
                reste -= pris
                if reste == 0:
                    break
            quantites[idx] = -reste

avant = len(df_clean)
df_clean["quantite"] = pd.Series(quantites)
df_clean = df_clean[df_clean["quantite"] != 0]
df_clean["montant_ligne"] = df_clean["quantite"] * df_clean["prix_unitaire"]

total_client = df_clean.groupby("id_client")["montant_ligne"].transform("sum")
df_clean = df_clean[total_client > 0].drop(columns="cle")

print(f"Lignes : {avant} avant, {len(df_clean)} après")
print(f"Clients : {df_clean['id_client'].nunique()}")
print(f"Lignes négatives restantes : {(df_clean['quantite'] < 0).sum()} pour {df_clean.loc[df_clean['quantite'] < 0, 'montant_ligne'].sum():.2f} £")
print(f"CA net : {df_clean['montant_ligne'].sum():.2f} £")
```

4 800 lignes ont disparu car leurs quantités sont tombées à 0 après la fusion. Il reste 388 557 lignes et 4 321 clients.

409 lignes restent négatives pour -7 719 £. Ce sont des retours plus grands que les achats du même article dans nos données. L'achat d'origine date d'avant décembre 2010. On les garde pour que le total de chaque client reste le même que dans le RFM. Pour compter seulement les ventes il faudra filtrer les quantités positives.

Le CA net est de 8 243 744 £. Il y a 288 £ d'écart avec la somme des montants du RFM car quelques retours ont été remboursés à un prix différent de celui de l'achat. Pendant la fusion on garde le prix d'achat. L'écart est négligeable (0,003 %).

## 4. Construction du dataset RFM

### 4.1 Date de référence

Pour calculer la Récence, il faut une date de départ. On prend le lendemain de la dernière facture, comme si on faisait l'analyse le 10 décembre 2011. C'est ce qui se fait en général en RFM. Ça décale juste tout le monde d'un jour, donc ça ne change rien aux écarts entre clients.

```python
# Date de référence : le lendemain de la dernière facture, à minuit (pour compter en jours entiers)
date_reference = df["date_facture"].max().normalize() + pd.Timedelta(days=1)
print("Date de référence :", date_reference)
```

### 4.2 Calcul de la Récence, la Fréquence et le Montant

On regroupe toutes les lignes par client. La Récence et la Fréquence se calculent seulement sur les achats (pas sur les retours). Le Montant, lui, se calcule sur toutes les lignes : les retours viennent se déduire des achats.

```python
# On regroupe par client pour calculer R, F et M
achats = df[~df["numero_facture"].str.startswith("C")]

# Récence et Fréquence : uniquement sur les achats
rfm = achats.groupby("id_client").agg(
    derniere_date=("date_facture", "max"),
    frequence=("numero_facture", "nunique")
)
rfm["recence"] = (date_reference - rfm["derniere_date"].dt.normalize()).dt.days

# Montant : sur toutes les lignes, retours compris
rfm["montant"] = df.groupby("id_client")["montant_ligne"].sum()

rfm = rfm[["recence", "frequence", "montant"]]
print(rfm.shape)
rfm.head()
```

On obtient un tableau de 4 324 clients, un par ligne, avec leur Récence, leur Fréquence et leur Montant. Par exemple, le client 12347 a acheté il y a 3 jours, a passé 7 commandes et a dépensé 4 310 £ au total.

### 4.3 Retrait des clients au montant nul ou négatif

Si le montant d'un client est à 0 ou en dessous, c'est qu'au final il n'a rien acheté (ses retours compensent ou dépassent ses achats). On les retire. En plus, le log de la partie 5 ne marche pas sur des valeurs négatives.

```python
# On regarde combien de clients ont un montant nul ou négatif, puis on les retire
print("Clients avec un montant <= 0 :", (rfm["montant"] <= 0).sum())

rfm = rfm[rfm["montant"] > 0]
print("Clients restants :", len(rfm))
```

Seulement 3 clients avaient un montant nul ou négatif : des cas où un retour porte sur un achat fait avant décembre 2010 qu'on ne voit pas dans nos données. On les retire, il reste 4 321 clients.

## 5. Traitement de l'asymétrie

### 5.1 Distributions avant transformation

```python
# Histogrammes de R, F et M pour voir leur forme, avec le coefficient d'asymétrie (skewness) de chacune
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, col in zip(axes, ["recence", "frequence", "montant"]):
    sns.histplot(rfm[col], bins=50, ax=ax)
    ax.set_title(f"{col} (skew = {rfm[col].skew():.2f})")

plt.tight_layout()
plt.show()
```

Les trois variables ont une longue traînée à droite. Pour la Récence ça reste correct (skew de 1,24) : beaucoup de clients récents puis une queue jusqu'à 374 jours.

Par contre la Fréquence (11,84) et surtout le Montant (21,65) sont très asymétriques. Presque tous les clients sont collés à gauche et quelques-uns sont super loin. On ne voit quasiment rien sur ces deux graphiques.

Pour K-Means qui calcule des distances, ces quelques clients écraseraient tout le reste. Donc on passe au log pour réduire l'asymétrie, puis on standardisera en partie 6 pour mettre les trois variables à la même échelle.

### 5.2 Transformation log1p

On applique `np.log1p` aux trois variables. On garde `rfm` tel quel (on en aura besoin pour interpréter les clusters avec les vraies valeurs) et on crée une version transformée `rfm_log`.

```python
# On applique log1p aux trois variables dans un nouveau tableau, rfm reste intact
rfm_log = np.log1p(rfm)
rfm_log.head()
```

Les valeurs sont maintenant beaucoup plus resserrées. Par exemple le client 12347 passe d'un montant de 4 310 £ à 8,37. Les écarts entre gros et petits clients existent toujours mais ils sont tassés.

Par contre les trois variables n'ont toujours pas la même échelle (le Montant tourne autour de 5 à 9 alors que la Fréquence est entre 0,7 et 2). C'est pour ça qu'on standardisera ensuite.

### 5.3 Distributions après transformation

```python
# Mêmes histogrammes qu'en 5.1 mais sur les données transformées, pour comparer le skew avant / après
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, col in zip(axes, ["recence", "frequence", "montant"]):
    sns.histplot(rfm_log[col], bins=50, ax=ax)
    ax.set_title(f"{col} (skew = {rfm_log[col].skew():.2f})")

plt.tight_layout()
plt.show()
```

Le log a fait son travail :
- Le Montant passe d'un skew de 21,65 à 0,36. On a maintenant presque une cloche.
- La Fréquence passe de 11,84 à 1,22. Elle reste un peu asymétrique à cause des 1 500 clients qui n'ont commandé qu'une fois : ils ont tous la même valeur, le log ne peut pas les étaler.
- La Récence passe de 1,24 à -0,33. Elle était déjà peu asymétrique, donc le log l'a fait passer un peu de l'autre côté, mais elle reste dans la zone "quasi symétrique" (entre -0,5 et 0,5).

Pour la Récence le log n'était pas indispensable. J'ai choisi de le faire quand même parce qu'elle est plus proche de la symétrie après (-0,33) qu'avant (1,24).

Les extrêmes ne sont plus à des kilomètres du reste. On peut passer à la standardisation.

## 6. Standardisation

### 6.1 Mise à l'échelle des variables

K-Means et la CAH calculent des distances entre clients. Si une variable a des valeurs plus grandes que les autres, elle pèse plus dans le calcul. On ramène donc les trois variables à la même échelle avec `StandardScaler` (moyenne 0, écart-type 1).

```python
# On standardise les trois variables (moyenne 0, écart-type 1) pour qu'elles pèsent pareil dans les distances
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
rfm_scaled = pd.DataFrame(
    scaler.fit_transform(rfm_log),
    index=rfm_log.index,
    columns=rfm_log.columns
)
rfm_scaled.head()
```

Les trois variables sont maintenant à la même échelle, autour de 0. Positif = au-dessus de la moyenne, négatif = en dessous.

Ex : le client 12347 a une Récence de -1,89, une Fréquence de 1,09 et un Montant de 1,45. Il achète souvent, beaucoup et récemment. Attention, pour la Récence c'est inversé : négatif veut dire qu'il a acheté plus récemment que la moyenne, donc c'est bien.

### 6.2 Vérification

```python
# On vérifie que chaque variable a bien une moyenne de 0 et un écart-type de 1
rfm_scaled.describe().round(2)
```

La moyenne est à 0 et l'écart-type à 1 pour les trois variables. Le -0.00 vient d'un arrondi.

La Fréquence monte jusqu'à 5,85. Les clients qui commandent beaucoup restent visibles mais beaucoup moins éloignés qu'avant le log. Le minimum et le premier quartile de la Fréquence sont identiques (-0,95). Ils correspondent aux clients qui n'ont commandé qu'une seule fois.

Le dataset est prêt pour le clustering.

## 7. Clustering K-Means

### 7.1 Méthode du coude

On lance K-Means pour K de 1 à 10 et on note l'inertie à chaque fois. L'inertie est la somme des distances entre chaque client et le centre de son cluster. Plus elle est basse, plus les clusters sont serrés. On cherche le point où elle arrête de baisser fortement, c'est le coude.

```python
# On teste K de 1 à 10 et on note l'inertie pour chaque K
from sklearn.cluster import KMeans

valeurs_k = range(1, 11)
inerties = []

for k in valeurs_k:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(rfm_scaled)
    inerties.append(kmeans.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(list(valeurs_k), inerties, marker="o")
plt.title("Méthode du coude")
plt.xlabel("Nombre de clusters K")
plt.ylabel("Inertie")
plt.show()
```

Le coude le plus net est à K = 2. L'inertie est divisée par deux entre K = 1 et K = 2. Elle baisse encore bien jusqu'à K = 3 puis de moins en moins. Après K = 4 ou 5 la courbe est presque plate.

Le coude n'est pas très franc après 2. On regarde aussi le score de silhouette pour choisir.

L'inertie à K = 1 vaut environ 12 963. C'est le nombre de clients (4 321) multiplié par le nombre de variables (3) car chaque variable standardisée a une variance de 1. 

### 7.2 Score de silhouette

Pour chaque client, la silhouette compare sa distance aux clients de son cluster avec sa distance aux clients du cluster voisin. Le score va de -1 à 1. Proche de 1, les clusters sont bien séparés. Proche de 0, ils se chevauchent. On garde le K qui a le meilleur score.

```python
# On teste K de 2 à 10 et on calcule le score de silhouette pour chaque K (il n'existe pas pour K = 1)
from sklearn.metrics import silhouette_score

valeurs_k_sil = range(2, 11)
silhouettes = []

for k in valeurs_k_sil:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(rfm_scaled)
    silhouettes.append(silhouette_score(rfm_scaled, labels))

plt.figure(figsize=(8, 4))
plt.plot(list(valeurs_k_sil), silhouettes, marker="o")
plt.title("Score de silhouette")
plt.xlabel("Nombre de clusters K")
plt.ylabel("Silhouette")
plt.show()

for k, s in zip(valeurs_k_sil, silhouettes):
    print(f"K = {k} : {s:.3f}")
```

Le meilleur score est à K = 2 (0,433). Ensuite le score baisse quand K augmente. K = 3 (0,340) et K = 4 (0,337) sont presque à égalité.
La silhouette ne monte pas jusqu'à un pic. Les clients ne forment pas des groupes bien séparés.
Il n'y a pas de nombre de groupes "naturel" dans les données. Le choix de K doit aussi tenir compte de l'utilité pour le marketing.

### 7.3 Comparaison de K = 3 et K = 4

Les deux valeurs ont presque la même silhouette. On lance les deux et on regarde la taille des clusters et leurs valeurs moyennes de Récence, Fréquence et Montant. On choisit celui qui donne les segments les plus utiles.

```python
# On lance K-Means avec K = 3 puis K = 4 et on affiche la taille et les moyennes RFM (vraies valeurs) de chaque cluster
for k in [3, 4]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(rfm_scaled)

    resume = rfm.assign(cluster=clusters).groupby("cluster").agg(
        nb_clients=("recence", "size"),
        recence=("recence", "mean"),
        frequence=("frequence", "mean"),
        montant=("montant", "mean")
    ).round(1)

    print(f"===== K = {k} =====")
    display(resume)
```

Avec K = 3 on obtient trois groupes. Les meilleurs clients (achat récent, beaucoup de commandes et gros montant), des clients moyens et des clients partis.

Avec K = 4 le groupe des meilleurs clients reste presque le même (730 et 715 clients). Le groupe des clients moyens est coupé en deux. D'un côté des clients récents avec peu d'achats qui ressemblent à des nouveaux clients. De l'autre de bons clients qui n'ont pas acheté depuis environ 70 jours et commencent à s'éloigner. Ce sont deux cibles très différentes pour le marketing.

On garde K = 4. La silhouette est presque la même qu'avec K = 3 (0,337 contre 0,340) et les segments sont plus utiles.

### 7.4 Test des paramètres

On fait varier les paramètres principaux de K-Means un par un pour voir leur effet. On garde K = 4.

`init` choisit comment placer les centres au départ. Soit `k-means++` qui les place loin les uns des autres soit `random` quiles place au hasard.

```python
# On compare les deux méthodes de placement des centres au départ avec un seul lancement pour bien voir la différence
import time

for methode in ["k-means++", "random"]:
    debut = time.time()
    kmeans = KMeans(n_clusters=4, init=methode, n_init=1, random_state=42)
    labels = kmeans.fit_predict(rfm_scaled)
    duree = time.time() - debut

    print(f"{methode} : inertie = {kmeans.inertia_:.1f} | silhouette = {silhouette_score(rfm_scaled, labels):.3f} | itérations = {kmeans.n_iter_} | durée = {duree:.2f} s")
```

`k-means++` et `random` donnent la même inertie (3 887,6). La silhouette est un peu différente (0,334 et 0,339) car quelques clients à la frontière changent de cluster. L'écart est minime.

La vraie différence est le nombre d'itérations. `random` a besoin de 28 tours contre 15 pour `k-means++`. Avec `k-means++` les centres partent déjà bien placés donc K-Means converge plus vite. Sur nos données ça ne change rien en temps mais sur un gros dataset ça compterait.

`n_init` est le nombre de fois où K-Means est relancé avec des départs différents. Il garde le lancement qui a la plus petite inertie.

```python
# On compare plusieurs nombres de relances avec un départ au hasard pour que l'effet soit visible
for n in [1, 5, 10, 20]:
    debut = time.time()
    kmeans = KMeans(n_clusters=4, init="random", n_init=n, random_state=42)
    labels = kmeans.fit_predict(rfm_scaled)
    duree = time.time() - debut

    print(f"n_init = {n} : inertie = {kmeans.inertia_:.1f} | silhouette = {silhouette_score(rfm_scaled, labels):.3f} | durée = {duree:.2f} s")
```

Plus on relance K-Means, plus l'inertie baisse un peu (de 3 887,6 à 3 887,1). Elle ne bouge plus entre 10 et 20 relances. 10 relances suffisent donc à trouver le meilleur résultat.

La silhouette baisse très légèrement quand l'inertie s'améliore. K-Means cherche la plus petite inertie et pas la meilleure silhouette. Les deux ne vont pas toujours dans le même sens mais l'écart est infime.

`max_iter` est le nombre maximum de tours autorisés. À chaque tour K-Means range les clients dans le cluster le plus proche puis recalcule les centres. Il s'arrête quand plus rien ne bouge ou quand il atteint `max_iter`.

```python
# On limite le nombre de tours pour voir ce qui se passe quand K-Means n'a pas le temps de converger
import warnings
warnings.filterwarnings("ignore")

for m in [2, 5, 10, 300]:
    kmeans = KMeans(n_clusters=4, init="random", n_init=1, max_iter=m, random_state=42)
    labels = kmeans.fit_predict(rfm_scaled)

    print(f"max_iter = {m} : inertie = {kmeans.inertia_:.1f} | silhouette = {silhouette_score(rfm_scaled, labels):.3f} | itérations faites = {kmeans.n_iter_}")
```

Avec 2, 5 ou 10 tours K-Means est arrêté avant d'avoir fini. L'inertie est plus haute et la silhouette plus basse. Plus on le laisse tourner, meilleur est le résultat.

Avec 300 tours autorisés, il s'arrête tout seul au bout de 28 car les centres ne bougent plus. `max_iter` sert de sécurité. Il doit juste être plus grand que le nombre de tours nécessaires. Au-delà sa valeur ne change rien.

On teste maintenant toutes les combinaisons des trois paramètres pour tenir compte des effets des uns sur les autres. On classe les combinaisons de la plus petite inertie à la plus grande.

```python
# On teste toutes les combinaisons de init, n_init et max_iter et on les classe par inertie
resultats = []

for methode in ["k-means++", "random"]:
    for n in [1, 5, 10, 20]:
        for m in [10, 50, 300]:
            kmeans = KMeans(n_clusters=4, init=methode, n_init=n, max_iter=m, random_state=42)
            labels = kmeans.fit_predict(rfm_scaled)
            resultats.append({
                "init": methode,
                "n_init": n,
                "max_iter": m,
                "inertie": round(kmeans.inertia_, 1),
                "silhouette": round(silhouette_score(rfm_scaled, labels), 3)
            })

grille = pd.DataFrame(resultats).sort_values("inertie")
grille
```

La meilleure combinaison est `k-means++` avec 20 relances (inertie de 3 887,0). L'écart avec les suivantes est minuscule.

Toutes les combinaisons avec seulement 10 tours sont en bas du tableau. 50 et 300 tours donnent exactement les mêmes résultats donc K-Means n'a jamais besoin de plus de 50 tours ici.

On voit l'effet des paramètres les uns sur les autres. La pire combinaison est `random` avec 1 seul lancement et 10 tours (3 946,4). Un mauvais départ sans relance et sans assez de tours donne un mauvais résultat. `k-means++` est beaucoup moins sensible au départ. Même avec 1 seul lancement il arrive à 3 887,6.

Dès que K-Means a assez de tours, toutes les combinaisons donnent entre 3 887,0 et 3 887,6. Sur nos données les paramètres changent très peu le résultat.

On garde `init="k-means++"`, `n_init=20` et `max_iter=300`. 300 sert de marge de sécurité et ne coûte rien car K-Means s'arrête tout seul quand il a fini.

## 8. Clustering hiérarchique (CAH)

### 8.1 Dendrogramme

La CAH part de chaque client tout seul. À chaque étape elle fusionne les deux groupes les plus proches jusqu'à n'avoir plus qu'un seul groupe. Le dendrogramme montre toutes ces fusions sous forme d'arbre. La hauteur d'une fusion indique à quel point les deux groupes étaient éloignés. On coupe l'arbre là où les fusions deviennent très hautes pour choisir K.

```python
# On construit l'arbre de la CAH avec la méthode de Ward et on affiche le dendrogramme
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

arbre = linkage(rfm_scaled, method="ward")

plt.figure(figsize=(14, 5))
dendrogram(arbre, truncate_mode="lastp", p=30, show_leaf_counts=True)
plt.title("Dendrogramme (méthode de Ward)")
plt.xlabel("Groupes de clients (nombre de clients entre parenthèses)")
plt.ylabel("Distance de fusion")
plt.show()
```

Le dendrogramme propose deux découpages intéréssants. K = 4 et K = 6 donc on compare les deux avant de choisir.

### 8.2 Comparaison de K = 4 et K = 6

On coupe l'arbre pour obtenir 4 puis 6 clusters. Pour chacun on regarde la silhouette, la taille des clusters et leurs moyennes de Récence, Fréquence et Montant avec les vraies valeurs.

```python
# On coupe l'arbre en 4 puis en 6 clusters et on compare silhouette, tailles et moyennes RFM
for k in [4, 6]:
    labels = fcluster(arbre, t=k, criterion="maxclust")

    resume = rfm.assign(cluster=labels).groupby("cluster").agg(
        nb_clients=("recence", "size"),
        recence=("recence", "mean"),
        frequence=("frequence", "mean"),
        montant=("montant", "mean")
    ).round(1)

    print(f"===== K = {k} | silhouette = {silhouette_score(rfm_scaled, labels):.3f} =====")
    display(resume)
```

K = 4 la CAH donne une élite de 330 clients, des fidèles, des clients récents qui achètent peu et un gros groupe de 1 616 clients partis.

K = 6 deux groupes restent identiques (les 330 de l'élite et les 981 récents qui achètent peu). Les deux autres sont coupés en deux.
- Les 1 616 partis deviennent 720 clients perdus (une seule commande il y a 255 jours en moyenne) et 896 clients à risque (2,3 commandes, dernier achat il y a 143 jours).
- Les 1 394 fidèles deviennent 276 clients très actifs (dernier achat il y a 3,8 jours) et 1 118 fidèles réguliers.

La silhouette est presque la même (0,271 et 0,268). K = 6 sépare les clients perdus des clients à risque, qui demandent des actions marketing différentes. Aucun groupe n'est trop petit (le plus petit fait 276 clients). On garde K = 6.

Le groupe des 276 très actifs a acheté en moyenne il y a moins de 4 jours. C'est le rush de début décembre. Ce segment est peut-être en partie un effet Noël.

### 8.3 Test des paramètres

Le paramètre principal de la CAH est la méthode de regroupement (`linkage`). Elle définit comment on mesure la distance entre deux groupes avant de les fusionner.
- `ward` fusionne les deux groupes qui font le moins augmenter la dispersion.
- `complete` utilise la distance entre les deux clients les plus éloignés des deux groupes.
- `average` utilise la distance moyenne entre tous les clients des deux groupes.
- `single` utilise la distance entre les deux clients les plus proches des deux groupes.

On compare les quatre avec K = 6.

```python
# On compare les quatre méthodes de regroupement avec K = 6 : silhouette et taille des clusters
for methode in ["ward", "complete", "average", "single"]:
    arbre_test = linkage(rfm_scaled, method=methode)
    labels = fcluster(arbre_test, t=6, criterion="maxclust")
    tailles = pd.Series(labels).value_counts().sort_index().tolist()

    print(f"{methode} : silhouette = {silhouette_score(rfm_scaled, labels):.3f} | tailles = {tailles}")
```

`single` a la meilleure silhouette (0,387) mais met 4 315 clients dans un seul groupe et laisse 5 clients isolés. Il fusionne deux groupes dès que deux de leurs clients sont proches. Ça crée une chaîne qui avale presque tout le monde. Les clients isolés sont très loin des autres et font monter la silhouette. Le score est bon mais la segmentation est inutilisable.

`average` fait presque pareil avec un groupe de 3 496 clients et des groupes de 2, 3 ou 10 clients.

`complete` donne des groupes plus équilibrés mais la silhouette la plus basse (0,195).

`ward` est la seule méthode qui donne des groupes équilibrés et cohérents. On la garde. Le choix de K = 6 fait sur son dendrogramme reste valable.

La silhouette seule ne suffit pas pour choisir. Il faut aussi regarder la taille des clusters.

### 9.4 Choix du modèle

La silhouette baisse naturellement quand K augmente. Pour comparer les deux modèles de façon juste, on calcule leur silhouette à K égal (4 et 6).

```python
# Silhouette des deux modèles à K = 4 et K = 6 pour les comparer à nombre de clusters égal
scores = []

for k in [4, 6]:
    labels_km = KMeans(n_clusters=k, init="k-means++", n_init=20, max_iter=300, random_state=42).fit_predict(rfm_scaled)
    labels_cah = AgglomerativeClustering(n_clusters=k, linkage="ward").fit_predict(rfm_scaled)
    scores.append({
        "K": k,
        "K-Means": round(silhouette_score(rfm_scaled, labels_km), 3),
        "CAH": round(silhouette_score(rfm_scaled, labels_cah), 3)
    })

pd.DataFrame(scores).set_index("K")
```

K-Means a une meilleure silhouette que la CAH à 4 comme à 6 clusters (0,338 contre 0,271 à K = 4 et 0,314 contre 0,268 à K = 6). Ses groupes sont mieux séparés dans les deux cas.

On a gardé la CAH à 6 clusters pour la richesse de ses segments. Comme K-Means à 6 a une meilleure silhouette, on regarde ses profils pour voir s'il trouve des segments aussi utiles.

```python
# Profils de K-Means à 6 clusters, avec les vraies valeurs, pour les comparer à ceux de la CAH
kmeans_6 = KMeans(n_clusters=6, init="k-means++", n_init=20, max_iter=300, random_state=42)
labels_km6 = kmeans_6.fit_predict(rfm_scaled)

lecture_km6 = rfm.assign(cluster=labels_km6).groupby("cluster").agg(
    nb_clients=("recence", "size"),
    recence=("recence", "mean"),
    frequence=("frequence", "mean"),
    montant=("montant", "mean")
).round(1)

lecture_km6["part_clients (%)"] = (lecture_km6["nb_clients"] / len(rfm) * 100).round(1)
lecture_km6.sort_values("montant", ascending=False)
```

K-Means à 6 retrouve les segments utiles de la CAH. Une élite (316 clients, 13 080 £), des clients à risque (929 clients, 117 jours), des perdus (965 clients, 235 jours) et des récents qui achètent peu (821 clients, 38 jours).

La différence est sur les fidèles. La CAH les coupe en 276 très actifs (dernier achat il y a 4 jours) et 1 118 fidèles. K-Means les coupe en 634 gros fidèles (7 commandes, 2 969 £) et 656 actifs récents (4 commandes, 1 206 £). Le découpage de K-Means est plus utile. Il sépare les fidèles selon leur volume d'achat plutôt que selon "ont-ils acheté cette semaine". Il dépend moins du rush de Noël.

K-Means à 6 a aussi une meilleure silhouette (0,314 contre 0,268) et des groupes plus équilibrés. On le retient comme modèle final.

```python
# On garde K-Means à 6 clusters comme modèle final et on ajoute son cluster dans rfm
modele_final = kmeans_6
rfm["cluster_final"] = labels_km6

print(rfm["cluster_final"].value_counts().sort_index())
```

Le modèle final est K-Means à 6 clusters. On retrouve les mêmes tailles que dans le tableau des profils. Le plus petit groupe fait 316 clients (l'élite) et le plus gros 965 (les perdus).

## 10. Réduction de dimension (PCA)

### 10.1 PCA et variance expliquée

On a 3 variables donc il faudrait un graphique en 3D pour voir tous les clients d'un coup. La PCA crée de nouveaux axes qui résument au mieux les 3 variables. On regarde quelle part de l'information chaque axe garde (la variance expliquée) pour savoir si 2 axes suffisent.

```python
# PCA sur les 3 variables standardisées : part de l'information gardée par chaque axe
from sklearn.decomposition import PCA

pca_complete = PCA(n_components=3)
pca_complete.fit(rfm_scaled)

variance = pd.DataFrame({
    "variance expliquée (%)": (pca_complete.explained_variance_ratio_ * 100).round(1),
    "cumul (%)": (pca_complete.explained_variance_ratio_.cumsum() * 100).round(1)
}, index=["Axe 1", "Axe 2", "Axe 3"])
variance
```

La variance mesure à quel point les clients sont différents les uns des autres. La variance expliquée d'un axe est la part de ces différences que l'axe arrive à montrer à lui seul.

L'axe 1 garde 75,4 % de l'information et l'axe 2 en ajoute 18,6 %. Avec ces deux axes on garde 94 % de l'information. L'axe 3 ne porte que 6 % et peut être laissé de côté. Un graphique en 2D sera donc très fidèle aux données.

### 10.2 Lecture des axes

Chaque axe de la PCA est un mélange des 3 variables. On regarde le poids de chaque variable dans les deux premiers axes pour comprendre ce qu'ils représentent.

```python
# Poids de chaque variable dans les deux premiers axes de la PCA
poids = pd.DataFrame(
    pca_complete.components_[:2],
    columns=rfm_scaled.columns,
    index=["Axe 1", "Axe 2"]
).round(2)
poids
```

L'axe 1 mélange les trois variables avec des poids proches. La Fréquence (0,62) et le Montant (0,60) d'un côté, la Récence (-0,51) de l'autre. À droite on trouve les clients qui achètent souvent, beaucoup et récemment. À gauche l'inverse. C'est un axe de valeur du client. Il porte 75 % de l'information car c'est la principale différence entre les clients.

L'axe 2 est dominé par la Récence (0,85) avec un peu de Montant (0,45). En haut on trouve des clients qui n'ont pas acheté depuis longtemps mais qui avaient un peu dépensé. En bas des clients récents. C'est surtout un axe d'ancienneté du dernier achat. Il permet de séparer des clients de même valeur qui n'ont pas acheté au même moment.

### 10.3 Visualisation des clusters en 2D

On projette les clients sur les deux premiers axes de la PCA et on les colore selon leur cluster final. Les croix noires sont les centres des clusters.

Rappel des clusters finaux. 0 = actifs récents, 1 = perdus, 2 = élite, 3 = à risque, 4 = récents qui achètent peu, 5 = gros fidèles.

Sur l'axe 1, de gauche à droite, on trouve les perdus, les récents qui achètent peu, les clients à risque, les actifs récents, les gros fidèles puis l'élite. C'est l'ordre de valeur qu'on avait dans les tableaux.

Les clients à risque sont juste au-dessus des récents qui achètent peu. Les gros fidèles sont juste au-dessus des actifs récents. Ils ont une valeur proche mais n'ont pas acheté au même moment.

L'élite est isolée à droite et très étalée. Ce sont quelques clients exceptionnels très différents les uns des autres.
